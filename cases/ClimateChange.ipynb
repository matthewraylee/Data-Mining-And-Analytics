# Case 1: Climate Change in Regression Analysis

```python
import pandas as pd
import statsmodels.formula.api as smf
import numpy as np
```

```python
climate_train = pd.read_csv('https://raw.githubusercontent.com/amandeep0/IS451/main/data/ClimateChangeTrain.csv')
climate_test = pd.read_csv('https://raw.githubusercontent.com/amandeep0/IS451/main/data/ClimateChangeTest.csv')
climate_train.head()
```

## a) Linear Regression Model to predict Temp, using all variables, except Year and Month

```python
linear_regression_model_1 = smf.ols('Temp ~ MEI + CO2 + CH4 + N2O + CFC11 + CFC12 + TSI + Aerosols', data=climate_train)
linear_regression_model_1_results = linear_regression_model_1.fit()
print(linear_regression_model_1_results.summary())
```

## a(i). Linear Regression Equation:

```python
# Get intercept and coefficients
intercept = linear_regression_model_1_results.params['Intercept']
coefficients = linear_regression_model_1_results.params.drop('Intercept')

# Generate Linear Regression Equation
equation1 = f"Temp = {intercept:.4f} + " + " + ".join([f"{coef:.4f}({name})" for name, coef in coefficients.items()])
print("Linear Regression Equation:\n", equation1)
```

## a(ii). Model Evaluation
$R^2$ = 0.751

Significant values (p-value < 0.05):
- MEI 
- CO2
- CFC11
- CHC12
- TSI
- Aerosols

## a(iii). Contradiction for N2O and CFC-11
The simplest explanation for this contradiction is due to multicollinearity that may be present with other variables (greehouse gasses). The model does not recognize individual effects towards other ones, and this results in the negative coefficient signs.

CFC.11 is highly correlated to:

## b) Linear Regression using only MEI, TSI, Aerosols, and N2O

```python
linear_regression_model_2 = smf.ols('Temp ~ MEI + TSI + Aerosols + N2O', data=climate_train)
linear_regression_model_2_results = linear_regression_model_2.fit()
print(linear_regression_model_2_results.summary())
```

### b(i) Coefficient Comparison to Previous Model

```python
### First Model
print("Previous Linear Regression Equation:\n", equation1)

### Second Model
# Get intercept and coefficients
intercept = linear_regression_model_2_results.params['Intercept']
coefficients = linear_regression_model_2_results.params.drop('Intercept')

# Generate Linear Regression Equation
equation2 = f"Temp = {intercept:.4f} + " + " + ".join([f"{coef:.4f}({name})" for name, coef in coefficients.items()])
print("Current Linear Regression Equation:\n", equation2)
```

The coefficient for N2O for previous model is -0.0165 while the new model is 0.0253. The N20 coefficient in the previous model is negative, suggesting a negative relationship with temperature, while the new model is positive, suggesting positive relationship with temperature. If we look further in the coefficients from the p-value, the previous model suggest that N20 is not really significant variable while in the new model, N20 is highly significant due to 0 p-value. 

In conclusion, in the previous model, N20 has weak negative relationship with relationship with temperature while in the new model, N20 is strongly significant variable such that the higher N20 level, the higher temperature level.

## c) Test model

```python
climate_test.head()
```

```python
prediction = linear_regression_model_2_results.predict(climate_test)
print(prediction)
```

```python
SSE = np.sum((prediction - climate_test['Temp'])**2)
SST = np.sum((climate_train['Temp'].mean() - climate_test['Temp'])**2)    

print("SSE:", SSE)
print("SST:", SST)

R2 = 1 - (SSE/SST)
print("R2:", R2)
```

$R^2$ on the test set is 0.4967794872202327, which is close to 0.5. Model is capturing moderate relationship between the independent variables and temperature. Still, more than 50% of variance in temperature could not be explained by this model, which means there are other factors that affects temperature. To improve this model, we can add more variables that are relevant, which may improve the R2 value.
